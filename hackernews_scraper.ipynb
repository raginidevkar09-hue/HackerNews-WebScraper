```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

base = "https://news.ycombinator.com/"
url = base + "news"

all_data = []

for page in range(6):  # ~180 stories
    r = requests.get(url)
    soup = BeautifulSoup(r.text, "html.parser")

    items = soup.find_all("tr", class_="athing")

    for item in items:
        try:
            title = item.find("span", class_="titleline").text
            link = item.find("a")["href"]
            item_id = item["id"]

            sub = item.find_next_sibling("tr").find("td", class_="subtext")

            author = sub.find("a", class_="hnuser").text if sub.find("a", class_="hnuser") else "N/A"
            score = sub.find("span", class_="score").text if sub.find("span", class_="score") else "0"
            time = sub.find("span", class_="age").text
            comments = sub.find_all("a")[-1].text

            all_data.append({
                "id": item_id,
                "title": title,
                "link": link,
                "author": author,
                "points": score,
                "comments": comments,
                "posted_time": time
            })
        except:
            pass

    next_page = soup.find("a", class_="morelink")["href"]
    url = base + next_page

df = pd.DataFrame(all_data)

df.to_csv("data.csv", index=False)
df.to_json("data.json", orient="records", indent=4)

df.head()
```
